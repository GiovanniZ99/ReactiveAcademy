##Indentazione

L'indentazione consiste negli spazi all'inizio di una riga di codice.

Mentre in altri linguaggi di programmazione l'indentazione nel codice è solo per la leggibilità, in Python l'indentazione è molto importante.

Python utilizza l'indentazione per indicare un blocco di codice.

```python
if 5 > 2:
  print("Five is greater than two!")
```

Avremo un errore se si salta l'indentazione, il numero di spazi è a discrezione del programmatore, basta che sia almeno uno e coerente per tutto il blocco

```python
if 5 > 2:
  print("Five is greater than two!")
if 5 > 2:
 print("Five is greater than two!")
if 5 > 2:
        print("Five is greater than two!")
        print("This is wrong")
```

### 📌 Metodi delle liste in Python

| Metodo | Descrizione | Esempio |
|--------|------------|---------|
| `append(x)` | Aggiunge un elemento alla fine della lista. | `lst.append(5)` |
| `extend(iterabile)` | Aggiunge tutti gli elementi di un iterabile (lista, tuple, set, ecc.) alla lista. | `lst.extend([6, 7, 8])` |
| `insert(pos, x)` | Inserisce l'elemento `x` alla posizione `pos`. | `lst.insert(2, 99)` |
| `remove(x)` | Rimuove la **prima occorrenza** dell'elemento `x`. | `lst.remove(3)` |
| `pop([pos])` | Rimuove e restituisce l'elemento in posizione `pos` (default: ultimo elemento). | `lst.pop(2)` |
| `clear()` | Rimuove tutti gli elementi dalla lista. | `lst.clear()` |
| `index(x, [start, end])` | Restituisce l'indice della **prima occorrenza** di `x`. | `lst.index(4)` |
| `count(x)` | Conta quante volte `x` appare nella lista. | `lst.count(2)` |
| `sort([key, reverse])` | Ordina la lista in ordine crescente (default). Se `reverse=True`, ordina in ordine decrescente. | `lst.sort()` o `lst.sort(reverse=True)` |
| `sorted(lst, key=...)` | Restituisce una nuova lista ordinata senza modificare l'originale. | `new_lst = sorted(lst)` |
| `reverse()` | Inverte l'ordine degli elementi della lista. | `lst.reverse()` |
| `copy()` | Restituisce una copia superficiale della lista. | `lst2 = lst.copy()` |


```python
# Creiamo una lista di esempio
lst = [3, 1, 4, 1, 5, 9, 2]

# Aggiungere un elemento alla fine
lst.append(6)
print(lst)  # [3, 1, 4, 1, 5, 9, 2, 6]

# Estendere con un'altra lista
lst.extend([7, 8])
print(lst)  # [3, 1, 4, 1, 5, 9, 2, 6, 7, 8]

# Inserire un elemento in una posizione specifica
lst.insert(2, 99)
print(lst)  # [3, 1, 99, 4, 1, 5, 9, 2, 6, 7, 8]

# Rimuovere un elemento
lst.remove(1)  # Rimuove la prima occorrenza di 1
print(lst)  # [3, 99, 4, 1, 5, 9, 2, 6, 7, 8]

# Ordinare la lista
lst.sort()
print(lst)  # [1, 2, 3, 4, 5, 6, 7, 8, 9, 99]

# Invertire la lista
lst.reverse()
print(lst)  # [99, 9, 8, 7, 6, 5, 4, 3, 2, 1]

# Ottenere una copia della lista
lst2 = lst.copy()
print(lst2)  # [99, 9, 8, 7, 6, 5, 4, 3, 2, 1]
```

```python
lst = [5, 2, 9, 1]
sorted_lst = sorted(lst)  # Crea una nuova lista ordinata
print(sorted_lst)  # [1, 2, 5, 9]
print(lst)  # [5, 2, 9, 1] (rimane invariata)

lst.sort()  # Ordina la lista originale
print(lst)  # [1, 2, 5, 9]
```
