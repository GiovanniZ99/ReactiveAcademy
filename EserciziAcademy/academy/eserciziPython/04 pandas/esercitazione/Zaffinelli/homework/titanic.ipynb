Import pandas

```python
import pandas as pd
```

Read the `titanic3.xls` file and set 'name' as the index

```python
tit = pd.read_excel('titanic3.xls', index_col='name')
tit
```

Drop the following columns: `['home.dest', 'embarked', 'sibsp', 'parch']`

```python
tit.drop(['home.dest', 'embarked', 'sibsp', 'parch'], axis=1, inplace=True)
```

Filter out all the passengers which did not survive.

```python
passengers_not_survived =tit[tit['survived'] == False]
passengers_not_survived
```

Considering that the **body** column is only used for passengers who did not survive, drop that column, and then do the same thing for the **survived** column.

```python
passengers_not_survived.drop(columns=['body'], inplace=True)
passengers_not_survived.drop(columns=['survived'], inplace=True)
```

What is the max **fare** in **3rd class**?

```python
tit_3_class = tit[tit['pclass'] == 3]
tit_3_class_max_fare = tit_3_class['fare'].max()
tit_3_class_max_fare
```

Rename the index name of 'name' to 'Passenger'

```python
tit.index.name = 'Passenger'
tit
```

**BONUS**

Create a new column that's called "Free Coffee", insert 1 if a passenger is either in 1st or 2nd class, otherwise insert 0

```python
tit['Free Coffe'] = 1
tit.loc[tit['pclass'] == 3, ['Free Coffe']] = 0
tit
```

Use value_counts() to verify that 3rd class doesn't get any free coffee

```python
tit_3_class = tit[tit['pclass'] == 3] 
tit_3_class['Free Coffe'].value_counts()
```
