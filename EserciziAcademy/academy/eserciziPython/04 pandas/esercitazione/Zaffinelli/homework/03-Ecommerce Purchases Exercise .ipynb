** Import pandas and read in the Ecommerce Purchases csv file and set it to a DataFrame called ecom. **

```python
import pandas as pd
```

```python
ecom = pd.read_csv('Ecommerce Purchases', delimiter=',')
```

**Check the head of the DataFrame.**

```python
ecom.head()
```

** How many rows and columns are there? **

```python
ecom.info()
```

** What is the average Purchase Price? **

```python
ecom_avg_purchase_price = ecom['Purchase Price'].mean()
ecom_avg_purchase_price
```

** What were the highest and lowest purchase prices? **

```python
ecom_highest_price = ecom['Purchase Price'].max()
ecom_highest_price
```

```python
ecom_lowest_price = ecom['Purchase Price'].min()
ecom_lowest_price
```

** How many people have English 'en' as their Language of choice on the website? **

```python
ecom_language_count = ecom.loc[ecom['Language'] == 'en'].shape[0]
ecom_language_count
```

** How many people have the job title of "Lawyer" ? **


```python
ecom_lawyer_jobtitle = ecom.loc[ecom['Job'] == 'Lawyer'].shape[0]
ecom_lawyer_jobtitle
```

** How many people made the purchase during the AM and how many people made the purchase during PM ? **

**(Hint: Check out [value_counts()](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.value_counts.html) ) **

```python
ecom_am_pm_count_purchase = ecom['AM or PM'].value_counts()
print(ecom_am_pm_count_purchase)
```

** What are the 5 most common Job Titles? **

```python
ecom_modes_jobtitles = ecom['Job'].value_counts().head(5)
ecom_modes_jobtitles
```

** Someone made a purchase that came from Lot: "90 WT" , what was the Purchase Price for this transaction? **

```python
# purchse_lot_90wt = ecom['Purchase Price'].loc(ecom['Lot'] == '90 WT')
lot_90wt = ecom.loc[ecom['Lot'] == '90 WT', 'Purchase Price']
lot_90wt
```

** What is the email of the person with the following Credit Card Number: 4926535242672853 **

```python
email_from_credit_card = ecom.loc[ecom['Credit Card'] == 4926535242672853, 'Email']
email_from_credit_card
```

** How many people have American Express as their Credit Card Provider *and* made a purchase above $95 ?**

```python
people_count_amexpr_and_95 = ecom.loc[(ecom['CC Provider'] == 'American Express') & (ecom['Purchase Price'] >= 95)].shape[0]
people_count_amexpr_and_95
```

** Hard: How many people have a credit card that expires in 2025? **

```python
people_count_ccexp_2025 = ecom.loc[ecom['CC Exp Date'].str.endswith('25') ].shape[0]
people_count_ccexp_2025
```

** Hard: What are the top 5 most popular email providers/hosts (e.g. gmail.com, yahoo.com, etc...) **

```python
popular_email_provider = ecom['Email'].str.split('@').str[1].value_counts().head(5)

popular_email_provider
```
