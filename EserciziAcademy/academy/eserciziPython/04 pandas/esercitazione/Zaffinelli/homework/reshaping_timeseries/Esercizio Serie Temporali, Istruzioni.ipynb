###### Il dataset contiene il numero di passi percorsi a partire da 01/01/2020 all' 11/05/2020 alle ore 15:52
    
- il numero di passi mostrato e' cumulativo per giornata, quindi ad ogni nuova giornata parte da zero e l'ultima osservazione di giornata (tipicamente alle 23:59:00) contiene il numero totale di passi percorsi in quel giorno

#### 1) Sposta a indice le informazioni relative al tempo


il risultato finale deve essere il seguente:
- index: contiene le informazioni relative al tempo (anno-mese-giorno ora-minuti-secondi)
- l'indice e' nominato "TEMPO"
- c'e' solo una colonna che contiene le info relative ai passi

```python
#passi['TEMPO'] = passi['Date'] + ' ' + passi['Time']
#passi.set_index(pd.to_datetime(passi['TEMPO']), inplace=True)
passi.drop(['Date', 'Time'], axis = 1,inplace=True)
```

##### 2) trova il giorno con il numero maggiore di passi
- ricorda che l'ultima osservazione di ogni giornata contiene il numero totale di passi per quel giorno

```python
max_passi = passi.resample('D').last()
max_passi.nlargest(1, 'Steps')
```

#### 3) Riordina il dataframe in base al numero totale di passi percorsi settimanalmente, a partire dalla settimana con piu' passi totali 


```python
max_passi.drop(['TEMPO'], axis=1, inplace=True)
passi_ordinati = max_passi.resample('W', kind='period').sum()
passi_ordinati.sort_values(by='Steps', ascending=False)
```

#### 4) parte 1: calcola il numero totale di passi percorsi in media ogni giorno per ogni mese
- l'output deve essere come il seguente:


```python
media_passi = max_passi.resample('M', kind='period').mean()
media_passi
```

 #### 4) parte 2: come punto sopra ma ogni mese deve essere indicato per nome come nell'output sotto
 - suggerimento: attenzione al parametro "kind" e alla trasformazione in DatetimeIndex o PeriodIndex

```python
media_passi.index= max_passi.resample('M').mean().index.month_name()
media_passi
```
