```python
#Esercizio 1: Leggere il file CSV e memorizzare i dati in una lista di dizionari
def leggi_csv(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        headers = lines[0].strip().split(';')
        dati = []
        for line in lines[1:]:
            values = line.strip().split(';')
            record = {headers[i]: values[i] for i in range(len(headers))}
            dati.append(record)
    return dati
#Esercizio 2: Contare il numero totale di incidenti
def conta_incidenti(dati):
    return sum(1 for record in dati if record['CLASSE ITIL'] == 'INCIDENT')
#Esercizio 3: Calcolare il tempo medio di gestione per tutti gli incidenti
def tempo_medio_gestione_incidenti(dati):
    tempi = [int(record['Tempo totale di gestione incident']) for record in dati]
    return sum(tempi) / len(tempi) if tempi else 0
#Esercizio 4: Contare il numero di incident che non rispettano il SLA (NO e N.A.)
def conta_incidenti_sla_no(dati):
    return sum(1 for record in dati if record['Rispetto SLA Competenza Gruppo'] in ['NO', 'N.A.'])
#Esercizio 5: Contare la percentuale degli incident che rispettano il SLA
def percentuale_incidenti_sla_si(dati):
    total_incidenti = len(dati)
    sla_si_count = sum(1 for record in dati if record['Rispetto SLA Competenza Gruppo'] == 'SI')
    return (sla_si_count / total_incidenti) * 100 if total_incidenti else 0
#Esercizio 6: Calcolare il tempo totale di gestione per tutti gli incident
def tempo_totale_gestione_incidenti(dati):
    return sum(int(record['Tempo totale di gestione incident']) for record in dati)
#Esercizio 7: Contare il numero di incident per ciascuna priorità di apertura
def conta_incidenti_per_priorita(dati):
    priorita_count = {}
    for record in dati:
        priorita = record['PRIORITA DI APERTURA INCIDENT']
        if priorita not in priorita_count:
            priorita_count[priorita] = 0
        priorita_count[priorita] += 1
    return priorita_count
#Esercizio 8: Calcolare il tempo medio di presa in carico per tutti gli incident
def tempo_medio_presa_in_carico_incidenti(dati):
    tempi = [int(record['Tempo medio di presa in carico']) for record in dati]
    return sum(tempi) / len(tempi) if tempi else 0
#Esercizio 9: Trovare l'incident con il tempo totale di gestione più lungo
def incidente_tempo_gestione_massimo(dati):
    max_tempo = -1
    max_record = None
    for record in dati:
        tempo = int(record['Tempo totale di gestione incident'])
        if tempo > max_tempo:
            max_tempo = tempo
            max_record = record
    return max_record
#Esercizio 10: Calcolare il numero medio di passaggi per tutti gli incident
def numero_medio_passaggi_incidenti(dati):
    passaggi = [int(record['Num. Passaggi']) for record in dati]
    return sum(passaggi) / len(passaggi) if passaggi else 0

#Esercizio 11: Verificare la formattazione della dati
import re
from datetime import datetime

def verifica_formattazione_dati(dati):
  pattern = re.compile(r'\d{2}/\d{2}/\d{4}')
  formato = True
  errori=0
  for record in dati:
    if 'Data creazione' in record:
      try:
        data_creazione = record['Data creazione']
        if not isinstance(data_creazione, datetime):
          raise ValueError
      except ValueError:
        data_creazione = record['Data creazione']
        if pattern.match(data_creazione):
          data_creazione = datetime.strptime(data_creazione, "%d/%m/%Y")
        else:
          errori += 1
      record['Data creazione'] = data_creazione
  return errori

  def correct_format(list):
    format_error = 0
    pattern = re.compile(r'\d{2}/\d{2}/\d{4}')
    for dict in list:
        if not pattern.match(str(dict["Data creazione"])):
            format_error += 1
    return format_error

#Esercizio 12: Contare il numero di incident per ciascun anno
def conta_incidenti_per_anno(dati):
    anno_count = {}
    for record in dati:
        if 'Data creazione' in record:
            anno = record['Data creazione'].year
            if anno not in anno_count:
                anno_count[anno] = 0
            anno_count[anno] += 1
    return anno_count

def count_incident_foreach_year(data_list):
  date_dict = {}
  for entry in data_list:
    date = entry["Data creazione"][-4:]
    if date not in date_dict:
      date_dict[date] = 0
    if entry["Classe ITIL"] == "INCIDENT":
      date_dict[date] += 1
  return date_dict

#Esercizio 13: Contare il numero di incident per ciascun mese
def conta_incidenti_per_mese(dati):
    mese_count = {}
    for record in dati:
        if 'Data creazione' in record:
            mese = record['Data creazione'].month
            if mese not in mese_count:
                mese_count[mese] = 0
            mese_count[mese] += 1
    return mese_count

#Esercizio 14: Trovare l'incidente più recente
def incidente_piu_recente(dati):
    max_data = datetime(1900,1,1)
    max_record = None
    for record in dati:
        if 'Data creazione' in record and record['Data creazione'] > max_data:
            max_data = record['Data creazione']
            max_record = record['ID INCIDENT']
    return max_record, max_data

print("Numero totale di incident:", conta_incidenti(dati))
print("Tempo medio di gestione per tutti gli incident:", tempo_medio_gestione_incidenti(dati))
print("Numero di incident che non rispettano il SLA (NO e N.A.):", conta_incidenti_sla_no(dati))
print("Percentuale degli incident che rispettano il SLA:", percentuale_incidenti_sla_si(dati))
print("Tempo totale di gestione per tutti gli incident:", tempo_totale_gestione_incidenti(dati))
print("Numero di incident per ciascuna priorità di apertura:", conta_incidenti_per_priorita(dati))
print("Tempo medio di presa in carico per tutti gli incident:", tempo_medio_presa_in_carico_incidenti(dati))
print("Incident con il tempo totale di gestione più lungo:", incidente_tempo_gestione_massimo(dati))
print("Numero medio di passaggi per tutti gli incident:", numero_medio_passaggi_incidenti(dati))
print("Formattazione della dati corretta:", verifica_formattazione_dati(dati))
print("Numero di incidenti per ciascun anno:", conta_incidenti_per_anno(dati))
print("Numero di incidenti per ciascun mese:", conta_incidenti_per_mese(dati))
print("Incidente più recente:", incidente_piu_recente(dati))
```
